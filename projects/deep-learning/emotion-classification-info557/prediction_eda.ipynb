# Prediction EDA — comparing CodaBench-scored submission against dev truth

- Empty rate: predicted vs true
- Per-class counts: predicted vs true
- Coverage % per class

```python
import pandas as pd
from collections import Counter

pred = pd.read_csv('prediction_result/submission.csv', keep_default_na=False)
dev = pd.read_csv('data/dev.csv', keep_default_na=False)

print(f'predictions: {len(pred)} rows')
print(f'dev truth:   {len(dev)} rows')
```

```python
# Empty rate comparison
pred_empty = (pred['labels'].str.strip() == '').sum()
dev_empty = (dev['labels'].str.strip() == '').sum()

print(f'predicted empty: {pred_empty}/{len(pred)} ({100 * pred_empty / len(pred):.1f}%)')
print(f'true empty:      {dev_empty}/{len(dev)} ({100 * dev_empty / len(dev):.1f}%)')
```

```python
# Per-class counts: predicted vs true
def count_labels(df):
    c = Counter()
    for s in df['labels']:
        c.update(s.split())
    return c

pred_counts = count_labels(pred)
true_counts = count_labels(dev)

all_classes = sorted(set(pred_counts) | set(true_counts))
table = pd.DataFrame({
    'true':      [true_counts.get(c, 0) for c in all_classes],
    'predicted': [pred_counts.get(c, 0) for c in all_classes],
}, index=all_classes)
table['coverage_%'] = (100 * table['predicted'] / table['true'].replace(0, pd.NA)).round(0)
table.sort_values('true', ascending=False)
```

```python
import pandas as pd
from collections import Counter

train = pd.read_csv('train.csv', keep_default_na=False)
dev = pd.read_csv('dev.csv', keep_default_na=False)
print(len(train), len(dev))
train.head()
```

```python
c = Counter()
for s in train['labels']:
    c.update(s.split())
pd.Series(c).sort_values(ascending=False)
```

```python
train['text'].str.split().apply(len).describe(percentiles=[.5, .9, .95, .99])
```

```python
toks = Counter()
for t in train['text']:
    toks.update(t.lower().split())
print('vocab:', len(toks))
```

```python
empty_train = (train['labels'] == '').mean()
empty_dev = (dev['labels'] == '').mean()
print(f'empty train: {empty_train:.1%}')
print(f'empty dev: {empty_dev:.1%}')
```

```python
print('dupes in train:', train['text'].duplicated().sum())
print('train texts also in dev:', train['text'].isin(dev['text']).sum())
```

```python
from itertools import combinations

pairs = Counter()
for s in train['labels']:
    labels = sorted(s.split())
    pairs.update(combinations(labels, 2))
pairs.most_common(10)
```
